```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Create a 2D grid of points
x = np.linspace(-2, 2, 9)
y = np.linspace(-2, 2, 9)
X, Y = np.meshgrid(x, y)
grid_points = np.vstack([X.ravel(), Y.ravel()])

# 2. Define Transformation Matrices
# Shear Matrix: Hat_i -> [1, 0], Hat_j -> [1, 1]
M_shear = np.array([[1, 1],
                    [0, 1]])

# Rotation Matrix (90° counter-clockwise): Hat_i -> [0, 1], Hat_j -> [-1, 0]
theta = np.pi / 2
M_rotation = np.array([[np.cos(theta), -np.sin(theta)],
                       [np.sin(theta),  np.cos(theta)]])

# 3. Apply Transformations using matrix multiplication (@ operator)
sheared_grid = M_shear @ grid_points
rotated_grid = M_rotation @ grid_points

# 4. Plotting
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))


# --- Plot 1: Original Grid ---
ax1.scatter(grid_points[0], grid_points[1], color='#3b82f6', alpha=0.7)
ax1.quiver(0, 0, 1, 0, angles='xy', scale_units='xy', scale=1, color='#ef4444', width=0.015, label=r'$\hat{i} = [1,0]$')
ax1.quiver(0, 0, 0, 1, angles='xy', scale_units='xy', scale=1, color='#10b981', width=0.015, label=r'$\hat{j} = [0,1]$')
ax1.set_title('1. Original Grid', fontsize=12, fontweight='bold')
ax1.set_xlim(-4, 4)
ax1.set_ylim(-4, 4)
ax1.grid(True, linestyle='--', alpha=0.4)
ax1.legend(loc='upper left')

# --- Plot 2: Sheared Grid ---
ax2.scatter(sheared_grid[0], sheared_grid[1], color='#8b5cf6', alpha=0.7)
trans_i_s = M_shear @ np.array([1, 0])
trans_j_s = M_shear @ np.array([0, 1])
ax2.quiver(0, 0, trans_i_s[0], trans_i_s[1], angles='xy', scale_units='xy', scale=1, color='#ef4444', width=0.015, label=r'Transformed $\hat{i} = [1,0]$')
ax2.quiver(0, 0, trans_j_s[0], trans_j_s[1], angles='xy', scale_units='xy', scale=1, color='#10b981', width=0.015, label=r'Transformed $\hat{j} = [1,1]$')
ax2.set_title('2. Sheared Grid\nMatrix: [[1, 1], [0, 1]]', fontsize=12, fontweight='bold')
ax2.set_xlim(-4, 4)
ax2.set_ylim(-4, 4)
ax2.grid(True, linestyle='--', alpha=0.4)
ax2.legend(loc='upper left')

# --- Plot 3: Rotated Grid ---
ax3.scatter(rotated_grid[0], rotated_grid[1], color='#f59e0b', alpha=0.7)
trans_i_r = M_rotation @ np.array([1, 0])
trans_j_r = M_rotation @ np.array([0, 1])
ax3.quiver(0, 0, trans_i_r[0], trans_i_r[1], angles='xy', scale_units='xy', scale=1, color='#ef4444', width=0.015, label=r'Transformed $\hat{i} = [0,1]$')
ax3.quiver(0, 0, trans_j_r[0], trans_j_r[1], angles='xy', scale_units='xy', scale=1, color='#10b981', width=0.015, label=r'Transformed $\hat{j} = [-1,0]$')
ax3.set_title('3. Rotated Grid (90° CCW)\nMatrix: [[0, -1], [1, 0]]', fontsize=12, fontweight='bold')
ax3.set_xlim(-4, 4)
ax3.set_ylim(-4, 4)
ax3.grid(True, linestyle='--', alpha=0.4)
ax3.legend(loc='upper left')

plt.tight_layout()
plt.show()
```
