```python
import matplotlib.pyplot as plt
import numpy as np

# Set up figure with two subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Define vectors v = [3, 2] and w = [4, 1]
v = np.array([3, 2])
w = np.array([4, 1])

# -------------------------------------------------------------
# PLOT 1: GEOMETRIC VIEW (Projection Shadow * Length of w)
# -------------------------------------------------------------
w_mag_sq = np.dot(w, w)
proj_v_on_w = (np.dot(v, w) / w_mag_sq) * w

# Plot vectors
ax1.quiver(0, 0, w[0], w[1], angles='xy', scale_units='xy', scale=1, color='#ef4444', width=0.012, label='Vector w = [4, 1]')
ax1.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='#3b82f6', width=0.012, label='Vector v = [3, 2]')

# Extended line along w
line_x = np.linspace(-1, 6, 100)
line_y = (w[1]/w[0]) * line_x
ax1.plot(line_x, line_y, color='#cbd5e1', linestyle='--', label='Line along w')

# Projection shadow vector (green)
ax1.quiver(0, 0, proj_v_on_w[0], proj_v_on_w[1], angles='xy', scale_units='xy', scale=1, color='#10b981', width=0.015, label='Projection of v onto w')

# Perpendicular drop line from v tip to projection tip
ax1.plot([v[0], proj_v_on_w[0]], [v[1], proj_v_on_w[1]], color='#8b5cf6', linestyle=':', linewidth=2, label='Perpendicular Drop')

# Text calculation box
dot_val = np.dot(v, w)
ax1.annotate(f'Dot Product v.w = {dot_val}\nProj Length * ||w|| = {np.linalg.norm(proj_v_on_w):.2f} * {np.linalg.norm(w):.2f} = {dot_val}', 
             (0.5, 3.0), fontsize=10, fontweight='bold', bbox=dict(boxstyle='round', facecolor='#f1f5f9', alpha=0.8))

ax1.set_xlim(-1, 6)
ax1.set_ylim(-1, 4)
ax1.set_title('1. GEOMETRIC VIEW\nProjection Length x Vector Length', fontsize=11, fontweight='bold')
ax1.grid(True, linestyle=':', alpha=0.5)
ax1.legend(loc='upper left')

# -------------------------------------------------------------
# PLOT 2: DUALITY VIEW (1x2 Matrix Transformation acting on v)
# -------------------------------------------------------------
ax2.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='#3b82f6', width=0.012, label='Input Vector v = [3, 2]')

# Show components being scaled by row matrix entries [4, 1]
ax2.plot([0, v[0]], [0, 0], color='#ef4444', linestyle='--', label='v_x = 3 (scaled by w_x = 4)')
ax2.plot([v[0], v[0]], [0, v[1]], color='#10b981', linestyle='--', label='v_y = 2 (scaled by w_y = 1)')

ax2.annotate('3 * 4 = 12', (1.5, -0.4), color='#ef4444', fontweight='bold', fontsize=10)

ax2.annotate('2 * 1 = 2', (3.1, 1.0), color='#10b981', fontweight='bold', fontsize=10)

ax2.annotate('1x2 Matrix: [4  1]\nCalculation: 4*(3) + 1*(2) = 14', 
             (0.5, 3.0), fontsize=10, fontweight='bold', 
             bbox=dict(boxstyle='round', facecolor='#f1f5f9', alpha=0.8))

ax2.set_xlim(-1, 6)
ax2.set_ylim(-1, 4)
ax2.set_title('2. DUALITY VIEW\n1x2 Matrix [4, 1] acting on Vector v', fontsize=11, fontweight='bold')
ax2.grid(True, linestyle=':', alpha=0.5)
ax2.legend(loc='upper left')

plt.tight_layout()
plt.show()
```
