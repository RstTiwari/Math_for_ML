```python
import matplotlib.pyplot as plt
import numpy as np

# Step 1: Define two linearly independent basis vectors
v = np.array([2, 1])   # Red vector
w = np.array([-1, 2])  # Green vector

# Step 2: Pick any two scalars (try changing these numbers!)
a = 2.5
b = 1.5

# Step 3: Compute the Linear Combination
v_scaled = a * v
w_scaled = b * w
target_vector = v_scaled + w_scaled

print(f"Scaled Vector v ({a} * v) : {v_scaled}")
print(f"Scaled Vector w ({b} * w) : {w_scaled}")
print(f"Target Combination (a*v + b*w): {target_vector}")

# --- Step 4: Plotting the Vectors and the Span ---
fig, ax = plt.subplots(figsize=(8, 8))

# Background: Generate a grid of points to show the SPAN (all reachable points)
grid_scalars = np.linspace(-3, 3, 11)
for s1 in grid_scalars:
    print(s1,"grid1")
    for s2 in grid_scalars:
        print(s2,"grid2")
        pt = s1 * v + s2 * w
        ax.scatter(pt[0], pt[1], color='#3b82f6', alpha=0.25, s=20)

# Vector 1 (Red Arrow): Moving along vector v
ax.quiver(0, 0, v_scaled[0], v_scaled[1], angles='xy', scale_units='xy', scale=1, 
          color='#ef4444', label=f'Step 1: {a} * v = [{v_scaled[0]}, {v_scaled[1]}]')

# Vector 2 (Green Arrow): Moving tip-to-tail along vector w
ax.quiver(v_scaled[0], v_scaled[1], w_scaled[0], w_scaled[1], angles='xy', scale_units='xy', scale=1, 
          color='#10b981', label=f'Step 2: {b} * w = [{w_scaled[0]}, {w_scaled[1]}]')

# Resultant Vector (Purple Arrow): Direct route from origin to target
ax.quiver(0, 0, target_vector[0], target_vector[1], angles='xy', scale_units='xy', scale=1, 
          color='#8b5cf6', label=f'Linear Combination = [{target_vector[0]}, {target_vector[1]}]')

# Mark the final target point
ax.scatter(target_vector[0], target_vector[1], color='#8b5cf6', s=80, zorder=5)

# Axis & Grid Setup
ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)
ax.set_xlim(-8, 8)
ax.set_ylim(-8, 8)
ax.set_xlabel('x-axis', fontsize=12)
ax.set_ylabel('y-axis', fontsize=12)
ax.set_title(f'Practicing Linear Combinations: {a}*v + {b}*w', fontsize=14, fontweight='bold')
ax.grid(True, linestyle='--', alpha=0.4)
ax.legend(loc='upper left', fontsize=10)

plt.tight_layout()
plt.show()
```
